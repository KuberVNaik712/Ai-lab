```python

def is_variable(x):
    return isinstance(x, str) and len(x) == 1 and x.islower()


def apply_substitution(term, subst):
    if isinstance(term, str):
        while term in subst:
            term = subst[term]
        return term

    return tuple(apply_substitution(x, subst) for x in term)


def occurs_check(var, term, subst):
    term = apply_substitution(term, subst)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs_check(var, x, subst) for x in term)

    return False


def unify(x, y, subst):
    x = apply_substitution(x, subst)
    y = apply_substitution(y, subst)

    if x == y:
        return subst

    if is_variable(x):
        if occurs_check(x, y, subst):
            return None
        subst[x] = y
        return subst

    if is_variable(y):
        if occurs_check(y, x, subst):
            return None
        subst[y] = x
        return subst

    if isinstance(x, tuple) and isinstance(y, tuple):
        if x[0] != y[0] or len(x) != len(y):
            return None

        for a, b in zip(x[1:], y[1:]):
            subst = unify(a, b, subst)

            if subst is None:
                return None

        return subst

    return None


# Main program
A = ("Eats", "x", "Apple")
B = ("Eats", "Riya", "y")

result = unify(A, B, {})

if result is None:
    print("Unification Failed")
else:
    print("Unification Successful")
    print("Substitution:", result)
    print("Unified Expression:", apply_substitution(A, result))
print("Name:Kuber V Naik")
print("USN:1WN24CS142")
```
