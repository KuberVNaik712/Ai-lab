```python
from heapq import heappush, heappop

initial = (2, 8, 3,
           1, 6, 4,
           7, 0, 5)

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)

def misplaced_tiles(state):
    count = 0
    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1
    return count

def manhattan_distance(state):
    distance = 0

    for i in range(9):
        if state[i] == 0:
            continue

        current_row = i // 3
        current_col = i % 3

        goal_position = goal.index(state[i])
        goal_row = goal_position // 3
        goal_col = goal_position % 3

        distance += abs(current_row - goal_row)
        distance += abs(current_col - goal_col)

    return distance

def get_neighbors(state):
    neighbors = []
    blank = state.index(0)

    row = blank // 3
    col = blank % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        new_row = row + dr
        new_col = col + dc

        if 0 <= new_row < 3 and 0 <= new_col < 3:
            new_blank = new_row * 3 + new_col

            new_state = list(state)
            new_state[blank], new_state[new_blank] = \
                new_state[new_blank], new_state[blank]

            neighbors.append(tuple(new_state))

    return neighbors

def a_star(heuristic):
    open_list = []

    g = 0
    h = heuristic(initial)
    f = g + h

    heappush(open_list, (f, g, initial))

    parent = {}
    cost = {initial: 0}

    while open_list:
        f, g, current = heappop(open_list)

        if current == goal:
            path = []

            while current in parent:
                path.append(current)
                current = parent[current]

            path.append(initial)
            path.reverse()

            return path

        for next_state in get_neighbors(current):
            new_g = g + 1

            if next_state not in cost or new_g < cost[next_state]:
                cost[next_state] = new_g

                h = heuristic(next_state)
                f = new_g + h

                heappush(open_list, (f, new_g, next_state))
                parent[next_state] = current

    return None

def print_board(state):
    for i in range(0, 9, 3):
        print(state[i], state[i+1], state[i+2])
    print()

print("CASE 1: MISPLACED TILES")

path = a_star(misplaced_tiles)

for state in path:
    print_board(state)

print("Number of moves:", len(path) - 1)

print("CASE 2: MANHATTAN DISTANCE")

path = a_star(manhattan_distance)

for state in path:
    print_board(state)

print("Number of moves:", len(path) - 1)

print("Kuber V Naik")
print("1WN24CS142")
```
