```python
import math

pruned = []

def alphabeta(node, alpha, beta, is_max):
    if not isinstance(node, list):
        return node

    val = -math.inf if is_max else math.inf

    for i, child in enumerate(node):
        child_val = alphabeta(child, alpha, beta, not is_max)
        if is_max:
            val = max(val, child_val)
            if val >= beta:
                pruned.append(node[i + 1:])
                break
            alpha = max(alpha, val)
        else:
            val = min(val, child_val)
            if val <= alpha:
                pruned.append(node[i + 1:])
                break
            beta = min(beta, val)

    print(f"{'MAX' if is_max else 'MIN'} -> Value: {val}, Alpha: {alpha}, Beta: {beta}")
    return val

# Game Tree from Slide 4
tree = [[[10, 9], [14, 18]], [[5, 4], [50, 3]]]

root_val = alphabeta(tree, -math.inf, math.inf, True)
print(f"\nFinal Root MAX Value: {root_val}")
print(f"Pruned Subtrees: {pruned}")
```
