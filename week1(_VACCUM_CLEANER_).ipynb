```python
def vacuum_cleaner_agent(location, room_status):

    print(f"Vacuum cleaner is in Room {location}")

    if room_status[location] == "Dirty":
        print(f"Cleaning Room {location}...")
        room_status[location] = "Clean"

    if location == "A":
        location = "B"
    else:
        location = "A"

    return location, room_status


location = "A"

room_status = {
    "A": "Dirty",
    "B": "Dirty"
}

print("Initial State:")
print(room_status)

while room_status["A"] == "Dirty" or room_status["B"] == "Dirty":
    location, room_status = vacuum_cleaner_agent(location, room_status)

print("\nFinal State:")
print(room_status)
print("Both rooms are clean!")

print("name: KUBER V NAIK")
print("USN: 1WN24CS142")
```
