```python
import random
import math

N = 8

# Calculate the number of attacking queen pairs
def cost(state):
    conflicts = 0

    for i in range(N):
        for j in range(i + 1, N):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1

    return conflicts


# Simulated Annealing algorithm
def simulated_annealing():
    current = [random.randint(0, N - 1) for _ in range(N)]
    T = 1000
    cooling_rate = 0.99

    while T > 0.1:
        if cost(current) == 0:
            return current

        # Generate a random neighbour
        next_state = current.copy()
        col = random.randint(0, N - 1)
        next_state[col] = random.randint(0, N - 1)

        delta_E = cost(current) - cost(next_state)

        # Accept better states or probabilistically accept worse states
        if delta_E > 0:
            current = next_state
        else:
            probability = math.exp(delta_E / T)

            if random.random() < probability:
                current = next_state

        # Decrease temperature
        T *= cooling_rate

    return current if cost(current) == 0 else None


# Display the solution
solution = simulated_annealing()

if solution is not None:
    print("Solution found:")
    for row in range(N):
        print(" ".join("Q" if solution[col] == row else "."
                       for col in range(N)))
    print("Cost:", cost(solution))
else:
    print("No solution found. Try again.")

print("Kuber V Naik")
print("1WN24CS142")
```
