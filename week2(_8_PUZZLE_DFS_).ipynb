```python
goal = [1, 2, 3,
        4, 5, 6,
        7, 8, 0]

def moves(state):
    result = []

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    directions = [
        (-1, 0),
        (1, 0),
        (0, -1),
        (0, 1)
    ]

    for dr, dc in directions:
        r = row + dr
        c = col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new_zero = r * 3 + c
            new_state = state.copy()

            new_state[zero], new_state[new_zero] = \
                new_state[new_zero], new_state[zero]

            result.append(new_state)

    return result


def IDS(start):
    depth = 0

    while True:
        print("Searching at depth:", depth)

        result = DLS(start, depth, [start])

        if result is not None:
            return result

        depth += 1


def DLS(state, depth, path):

    if state == goal:
        return path

    if depth == 0:
        return None

    for child in moves(state):

        if child not in path:
            result = DLS(child, depth - 1, path + [child])

            if result is not None:
                return result

    return None


def display(state):
    print(state[0], state[1], state[2])
    print(state[3], state[4], state[5])
    print(state[6], state[7], state[8])
    print()


start = [1, 2, 3,
         4, 0, 6,
         7, 5, 8]

path = IDS(start)

print("\nGoal found!")
print("Moves:", len(path) - 1)

for state in path:
    display(state)

print("Kuber V Naik")
print("USN: 1WN24CS142")
```
