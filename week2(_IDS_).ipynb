```python
start = (5, 4, 0, 6, 1, 8, 7, 3, 2)
goal = (0, 1, 2, 3, 4, 5, 6, 7, 8)

def dfs(s, d, path):
    if s == goal:
        return path

    if d == 0:
        return None

    z = s.index(0)
    r, c = divmod(z, 3)

    for dr, dc in [(0,1),(1,0),(0,-1),(-1,0)]:
        nr, nc = r + dr, c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            nz = nr * 3 + nc

            x = list(s)
            x[z], x[nz] = x[nz], x[z]
            x = tuple(x)

            if x not in path:
                ans = dfs(x, d - 1, path + [x])

                if ans:
                    return ans

    return None


for depth in range(1, 30):
    ans = dfs(start, depth, [start])

    if ans:
        print("Moves:", len(ans) - 1)

        for s in ans:
            print(s[:3])
            print(s[3:6])
            print(s[6:])
            print()

        break

print("Kuber V Naik")
print("USN: 1WN24CS142")
```
