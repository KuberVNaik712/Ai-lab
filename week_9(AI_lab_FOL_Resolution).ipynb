```python

def resolve(c1, c2):
    for l1 in c1:
        for l2 in c2:
            if l1 == "NOT " + l2 or l2 == "NOT " + l1:
                new_clause = list(set(c1 + c2))
                new_clause.remove(l1)
                new_clause.remove(l2)
                return new_clause
    return None


def resolution(KB):
    while True:
        new_clauses = []

        for i in range(len(KB)):
            for j in range(i + 1, len(KB)):
                result = resolve(KB[i], KB[j])

                if result is not None:
                    if len(result) == 0:
                        return True

                    if result not in KB and result not in new_clauses:
                        new_clauses.append(result)

        if not new_clauses:
            return False

        KB.extend(new_clauses)


KB = [
    ["Eats(Anil,Peanuts)"],
    ["Alive(Anil)"],
    ["NOT Alive(Anil)", "NotKilled(Anil)"],
    ["NOT Eats(Anil,Peanuts)", "NOT NotKilled(Anil)", "Food(Peanuts)"],
    ["NOT Food(Peanuts)", "Likes(John,Peanuts)"],
    ["NOT Likes(John,Peanuts)"]
]

if resolution(KB):
    print("John likes peanuts is proved.")
else:
    print("Proof failed.")
```
